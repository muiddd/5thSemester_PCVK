# **Pambahasan Pertanyaan Praktikun D3**
1. **Apakah perbedaan gambar yang ditampilkan tanpa dan dengan matplotlib?** Perbedaan utamanya terletak pada tampilan koordinat piksel dan format warna. Menampilkan gambar tanpa matplotlib (misalnya menggunakan cv2_imshow dari OpenCV) akan menampilkan gambar secara langsung sesuai aslinya. Sedangkan, menampilkan gambar dengan Matplotlib (plt.imshow) akan memasukkan gambar ke dalam plot yang dilengkapi dengan sumbu koordinat berupa angka, sehingga kita bisa melihat ukuran panjang dan lebar piksel gambar tersebut. Selain itu, karena OpenCV membaca gambar dalam format warna BGR (Blue-Green-Red), jika gambar ditampilkan menggunakan Matplotlib tanpa dikonversi ke RGB terlebih dahulu, warnanya akan terlihat berbeda (tertukar) pada plotter.
2. **Apakah perbedaan dan pengaruhnya pembuatan black image antara tipe data
int16 dan int32?** Perbedaan tersebut berkaitan dengan presisi bit integer dan kompatibilitas dengan fungsi-fungsi spesifik di OpenCV. Pembuatan black image dengan tipe data int16 dapat digunakan untuk menggambar bentuk geometri dasar seperti persegi panjang atau lingkaran. Namun, pembuatan array dengan tipe data int32 secara khusus sangat dibutuhkan dan berpengaruh ketika kita mendefinisikan array titik koordinat (vertices) untuk menggambar bentuk polyline (garis banyak yang saling terhubung) menggunakan fungsi cv.polylines
3. **Apakah kegunaan “google.colab.patches import cv2_imshow” pada potongan
kode berikut** Kegunaannya adalah untuk mengimpor fungsi khusus dari Google Colab agar kita dapat menampilkan citra (gambar) secara langsung (image display) di dalam notebook browser. Fungsi penampil gambar asli bawaan OpenCV yaitu cv2.imshow() tidak didukung untuk dijalankan secara langsung di antarmuka Google Colab, sehingga harus digantikan dengan cv2_imshow().
4. **Apakah kegunaan “skimage import io” pada potongan kode soal nomor 3** Kegunaannya adalah untuk mengimpor modul io (input/output) dari pustaka scikit-image (skimage) yang berfungsi khusus untuk membaca file gambar. Pada praktikum ini, fungsi io.imread(url) sangat berguna karena memungkinkan program untuk membaca dan mengambil gambar langsung dari sebuah alamat tautan internet (URL) tanpa perlu mengunduhnya terlebih dahulu.


# **Tugas Praktikum D3**




```python
import cv2 as cv
import numpy as np
import random
from google.colab.patches import cv2_imshow

# Pastikan file KTM.jpeg sudah diunggah/di-load di Colab kamu
file_nama = "KTM.jpeg"
img_bgr = cv.imread(file_nama)

if img_bgr is None:
    print(f"Gambar {file_nama} tidak ditemukan. Pastikan kamu sudah melakukan upload/mount gdrive!")
else:
    # Mengambil dimensi gambar
    tinggi = img_bgr.shape[0]
    lebar = img_bgr.shape[1]

    # =============================================================
    # TUGAS BAGIAN 1 & 2
    # =============================================================

    # --- TUGAS 2: Tampilkan image channel Red-Blue dan Green-Blue ---
    img_rb = img_bgr.copy()
    img_rb[:, :, 1] = 0  # Matikan channel Green (Indeks 1)

    img_gb = img_bgr.copy()
    img_gb[:, :, 2] = 0  # Matikan channel Red (Indeks 2)

    print("Tugas 2: Gambar Red-Blue (Kiri) & Green-Blue (Kanan)")
    cv2_imshow(cv.hconcat((img_rb, img_gb)))

    # --- TUGAS 3: Tampilkan baris ke 20-115, kolom 25-120 ---
    img_crop1 = img_bgr[20:115, 25:120]
    print("\nTugas 3: Potongan Baris 20-115, Kolom 25-120")
    cv2_imshow(img_crop1)

    # --- TUGAS 4: Tampilkan baris 5-30, semua kolom, channel Red saja ---
    # Karena BGR, Red ada di indeks ke-2
    img_crop2 = img_bgr[5:30, :, 2]
    print("\nTugas 4: Baris 5-30, Semua Kolom, Channel Red (Ditampilkan sbg Grayscale)")
    cv2_imshow(img_crop2)

    # --- TUGAS 5: Buat 5 kotak berbagai ukuran dan warna acak ---
    img_kotak = img_bgr.copy()
    for _ in range(5):
        # Titik acak yang dijaga agar tidak keluar dari dimensi gambar
        x1 = random.randint(0, max(1, lebar - 100))
        y1 = random.randint(0, max(1, tinggi - 100))
        x2 = x1 + random.randint(20, 100)
        y2 = y1 + random.randint(20, 100)

        # Warna acak BGR (0-255)
        warna_acak = (random.randint(0,255), random.randint(0,255), random.randint(0,255))
        cv.rectangle(img_kotak, (x1, y1), (x2, y2), warna_acak, 3)

    print("\nTugas 5: 5 Kotak Acak")
    cv2_imshow(img_kotak)

    # --- TUGAS 6: Tampilkan image dengan posisi terbalik ---
    # cv.flip dengan kode 0 membalik gambar secara vertikal (atas-bawah)
    img_terbalik = cv.flip(img_bgr, 0)
    print("\nTugas 6: Gambar Posisi Terbalik")
    cv2_imshow(img_terbalik)


    # =============================================================
    # TUGAS BAGIAN 3 & 4
    # =============================================================

    # --- TUGAS 7: Rectangle & Circle di bagian wajah pada gambar ---
    img_wajah = img_bgr.copy()

    # PERKIRAAN POSISI: Karena layout KTM biasanya fotonya ada di sisi kiri,
    # kita set perkiraan koordinatnya. Jika posisinya meleset dari wajah di KTM-mu,
    # kamu bisa ubah nilai variabel titik_wajah_x dan titik_wajah_y di bawah ini.
    titik_wajah_x = int(lebar * 0.15) # Perkiraan posisi x (kiri)
    titik_wajah_y = int(tinggi * 0.5) # Perkiraan posisi y (tengah)
    radius_wajah = 50

    # Lingkaran Biru
    cv.circle(img_wajah, (titik_wajah_x, titik_wajah_y), radius_wajah, (255, 0, 0), 3)
    # Kotak Hijau
    cv.rectangle(img_wajah, (titik_wajah_x-radius_wajah, titik_wajah_y-radius_wajah),
                 (titik_wajah_x+radius_wajah, titik_wajah_y+radius_wajah), (0, 255, 0), 3)

    print("\nTugas 7: Kotak & Lingkaran pada Wajah KTM")
    cv2_imshow(img_wajah)

    # --- TUGAS 8 & 9: Channel B, Kotak Kiri Bawah, dan Nama File ---
    img_b_only = img_bgr.copy()

    # Untuk mendapatkan Channel Blue (indeks 0), matikan Green (1) dan Red (2)
    img_b_only[:, :, 1] = 0
    img_b_only[:, :, 2] = 0

    # Tugas 8: Kotak di sudut kiri bawah
    cv.rectangle(img_b_only, (0, tinggi-80), (150, tinggi), (255, 255, 255), 3)

    # Tugas 9: Teks nama file
    cv.putText(img_b_only, file_nama, (10, tinggi-30), cv.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)

    print(f"\nTugas 8 & 9: Channel Blue, Kotak di Kiri Bawah & Teks '{file_nama}'")
    cv2_imshow(img_b_only)
```
